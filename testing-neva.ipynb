```python
%%writefile prompts.py

"""
Prompt bank for SDXL LoRA style transfer evaluation.

500 prompts for SDXL LoRA style transfer evaluation.

Categories cover a wide range of visual concepts to stress-test
whether the LoRA style (Gris/Neva watercolor game art) transfers
consistently across different content types.

Usage:
    from prompts import BASE_PROMPTS, get_lora_prompts, get_vanilla_prompts, get_clip_prompts
"""

# =============================================================================
# BASE PROMPTS (no style tokens — content only)
# =============================================================================
# These are the "neutral" prompts. The generation script will:
#   - LoRA:    prepend "<s0><s1> style" to each
#   - Vanilla: prepend "painterly style" to each
# =============================================================================

BASE_PROMPTS = [
    # =========================================================================
    # 1. NATURAL ENVIRONMENTS (50)
    # =========================================================================
    "a misty forest at dawn with sunlight filtering through tall pines",
    "a mountain lake reflecting snow-capped peaks under a clear sky",
    "a winding river through an autumn valley with golden trees",
    "a dense bamboo grove with soft diffused green light",
    "a rocky coastline with crashing waves at sunset",
    "a meadow full of wildflowers under a dramatic cloudy sky",
    "a frozen waterfall in a winter canyon",
    "a tropical rainforest canopy seen from below",
    "a volcanic landscape with steam rising from black rock",
    "a rolling hillside covered in lavender under golden hour light",
    "a birch forest in early spring with new green leaves",
    "a desert canyon at twilight with purple shadows",
    "a mossy old-growth forest with massive tree roots",
    "a glacier flowing into a turquoise fjord",
    "a savanna with scattered acacia trees at dusk",
    "a rice paddy terrace on a misty mountainside",
    "a coral reef seen from above through crystal clear water",
    "a redwood forest with fog drifting between enormous trunks",
    "a snowy tundra under the northern lights",
    "a marshland with tall reeds and a heron standing still",
    "a cherry blossom grove in full bloom with petals falling",
    "a cliffside waterfall plunging into a jungle pool",
    "a wheat field stretching to the horizon under stormy skies",
    "a mangrove swamp at low tide with tangled roots exposed",
    "a sand dune field with rippled patterns from the wind",
    "a pine forest blanketed in fresh snow at night",
    "a river delta seen from above with branching waterways",
    "a cave interior with stalactites and an underground lake",
    "an alpine meadow with wildflowers and distant peaks",
    "a kelp forest underwater with shafts of light from above",
    "a foggy moor with heather and a lone standing stone",
    "a sunflower field under a bright blue summer sky",
    "a slot canyon with narrow walls and light beams from above",
    "a floating bog with cotton grass and dark peat water",
    "a coastal dune with sea grass blowing in the wind",
    "a temperate rainforest with ferns and fallen logs",
    "a crater lake with impossibly blue water",
    "a dry riverbed with cracked earth and scattered stones",
    "a palm oasis in the middle of a vast desert",
    "a mountain pass with prayer flags and distant valleys",
    "a tide pool with colorful anemones and starfish",
    "a dormant volcano crater with moss growing inside",
    "a bioluminescent bay glowing blue at night",
    "a petrified forest with ancient stone trees",
    "a flooded forest with trees standing in still water",
    "a windy plateau with bent trees and low clouds",
    "a hot spring pool with steam and mineral deposits",
    "a river gorge with rapids and autumn foliage",
    "a lichen-covered rocky shoreline at low tide",
    "a cloud forest with orchids and hanging moss",

    # =========================================================================
    # 2. URBAN & ARCHITECTURE (50)
    # =========================================================================
    "a narrow European cobblestone alley with hanging laundry",
    "a futuristic city skyline at night with glowing towers",
    "a ruined temple overgrown with vines and roots",
    "an old wooden bridge over a canal in a quiet village",
    "a grand cathedral interior with stained glass light",
    "a bustling open-air market with colorful awnings",
    "a rooftop garden overlooking a sprawling city",
    "a lighthouse on a rocky promontory in a storm",
    "an abandoned train station with overgrown tracks",
    "a Japanese torii gate at the entrance to a forest shrine",
    "a Venetian canal with gondolas and crumbling facades",
    "a modernist glass building reflecting the sunset",
    "a medieval castle perched on a cliff edge",
    "a colorful row of townhouses on a steep hill",
    "a subway platform with dramatic fluorescent lighting",
    "an ancient Roman aqueduct stretching across a valley",
    "a floating village with houses on stilts over water",
    "a brutalist concrete building against a grey sky",
    "a pagoda on a hillside surrounded by autumn maples",
    "a winding mountain road with hairpin turns",
    "a greenhouse interior full of tropical plants",
    "a stone amphitheater in a sun-drenched landscape",
    "a suspension bridge disappearing into fog",
    "a narrow canal in Amsterdam with bicycles on the bridge",
    "a crumbling brick factory with broken windows",
    "a whitewashed village on a Greek island hillside",
    "a train crossing a tall viaduct over a deep valley",
    "a traditional wooden house in a snowy mountain setting",
    "a spiral staircase seen from above",
    "a gothic arched corridor with dramatic shadows",
    "a neon-lit street in a rain-soaked city",
    "a windmill on a flat landscape with dramatic clouds",
    "a stone tower on an isolated island",
    "a grand library with floor-to-ceiling bookshelves",
    "a covered wooden bridge over a rushing stream",
    "a modern art museum with a sweeping curved roof",
    "a bazaar corridor with hanging lamps and textiles",
    "a clock tower in a small town square",
    "a garden maze seen from above",
    "an observatory dome on a mountaintop at twilight",
    "a thatched-roof cottage with a flower garden",
    "a concrete dam holding back a vast reservoir",
    "a mosque dome and minarets silhouetted at sunset",
    "a boardwalk along a beach with wooden railings",
    "a fire escape zigzagging down a brick building",
    "a drawbridge partially raised over a river",
    "a stone well in the center of a village courtyard",
    "a highway overpass with streaks of car lights at night",
    "a treehouse built high in an ancient oak",
    "a traditional courtyard house with a central fountain",

    # =========================================================================
    # 3. CHARACTERS & FIGURES (50)
    # =========================================================================
    "a solitary figure walking along a beach at sunset",
    "a woman reading under a large tree in a park",
    "a child flying a kite on a windy hilltop",
    "a hooded traveler approaching a distant walled city",
    "an old man sitting on a bench feeding pigeons",
    "a dancer mid-leap in an empty theater",
    "a knight on horseback crossing a shallow river",
    "a woman standing at a rain-streaked window looking out",
    "a fisherman casting a line from a rocky shore at dawn",
    "a group of monks walking through a misty forest path",
    "a young girl reaching toward a butterfly in a meadow",
    "a figure with an umbrella walking through a downpour",
    "a musician playing cello alone in a cathedral",
    "a shepherd with a flock on a green hillside",
    "a woman sitting on a dock with her feet in the water",
    "a blacksmith working at a forge with sparks flying",
    "a small child looking up at a towering ancient tree",
    "a figure standing at the edge of a vast canyon",
    "a woman with flowing hair standing in a strong wind",
    "a samurai walking through a bamboo grove",
    "a painter at an easel in a sunlit studio",
    "a figure in a small boat on a vast calm sea",
    "a grandmother knitting by a fireplace",
    "a warrior with a shield facing a coming storm",
    "a woman picking flowers in a cottage garden",
    "a lone rider on a horse in an open prairie",
    "a figure meditating on a rock overlooking a valley",
    "a man carrying a lantern through a dark forest",
    "a girl sitting on a crescent moon in a starry sky",
    "a figure climbing a winding mountain stairway",
    "a diver descending into deep blue water",
    "a woman playing violin on a rooftop at dusk",
    "a wanderer resting under a desert rock formation",
    "a child building a sandcastle on a quiet beach",
    "a figure walking through a field of tall golden wheat",
    "a woman sitting in a rowboat among water lilies",
    "a traveler with a backpack looking at a map",
    "a figure silhouetted against a massive full moon",
    "a young woman with a white wolf companion in a forest",
    "a person sleeping under a tree with a book on their chest",
    "a figure releasing a paper lantern into the night sky",
    "a swordsman standing on a bridge in the rain",
    "a woman tending a rooftop garden in a city",
    "a figure standing in ankle-deep water on a salt flat",
    "a child chasing fireflies at twilight",
    "a woman wrapped in a shawl standing on a snowy bridge",
    "a figure looking up at an enormous tree of light",
    "a person sketching in a cafe by the window",
    "a figure walking away down a long empty road",
    "a woman holding a glowing orb in a dark forest",

    # =========================================================================
    # 4. ANIMALS & CREATURES (50)
    # =========================================================================
    "a white fox sitting in a snowy landscape",
    "a great blue heron standing in shallow water at sunrise",
    "a pod of whales seen from above in deep blue ocean",
    "a deer standing at the edge of a foggy forest clearing",
    "an owl perched on a branch in moonlight",
    "a bear catching salmon in a rushing river",
    "a flock of starlings forming a murmuration at dusk",
    "a lone wolf howling on a rocky outcrop",
    "a sea turtle gliding through clear turquoise water",
    "a cat sleeping in a patch of afternoon sunlight",
    "butterflies swarming around a flowering bush",
    "a horse galloping through shallow ocean surf",
    "a crane standing on one leg in a misty pond",
    "a jellyfish floating in dark deep water with soft glow",
    "a pair of swans on a glass-still lake",
    "a fox kit playing in autumn leaves",
    "an eagle soaring above a mountain range",
    "a school of colorful fish on a coral reef",
    "a rabbit sitting in a moonlit garden",
    "a chameleon on a branch with vivid colors",
    "a polar bear on an ice floe under grey skies",
    "a peacock displaying its full tail feathers",
    "koi fish swimming in a clear garden pond",
    "a moth resting on a window at night",
    "a dragonfly hovering over a lily pad",
    "a stag with large antlers in morning mist",
    "a gecko climbing a rain-soaked leaf",
    "a flock of flamingos in a shallow pink lake",
    "a sleeping cat curled on a stack of old books",
    "a whale breaching out of dark ocean water",
    "an octopus camouflaged against a rocky seabed",
    "a hummingbird feeding from a red flower",
    "a snow leopard on a high mountain ledge",
    "a colony of penguins on an icy shore",
    "a spider web with morning dew drops",
    "a wild horse herd running across a steppe",
    "a kingfisher diving into a stream",
    "a tortoise walking slowly across a sandy path",
    "a raven perched on a bare winter branch",
    "a firefly swarm illuminating a forest glade",
    "a manatee floating in murky green water",
    "a falcon diving through clouds",
    "a hermit crab on a white sand beach",
    "a cheetah resting in tall golden grass",
    "a frog sitting on a mushroom in the rain",
    "a caterpillar on a leaf with tiny dewdrops",
    "a dolphin leaping out of sunset-lit waves",
    "a snail crossing a wet mossy stone",
    "a red panda sleeping in a tree fork",
    "a moth swarm around a single streetlamp at night",

    # =========================================================================
    # 5. STILL LIFE & OBJECTS (50)
    # =========================================================================
    "a cup of tea on a windowsill with rain outside",
    "an old leather-bound book on a wooden desk",
    "a vase of sunflowers on a white tablecloth",
    "a pair of worn boots by a cottage door",
    "a candle burning in a dark room with long shadows",
    "an hourglass with sand halfway through",
    "a typewriter with a half-written page",
    "a basket of freshly picked apples",
    "a compass and old map spread on a table",
    "a violin resting on a velvet chair",
    "a bowl of ripe peaches on a marble countertop",
    "a lantern hanging from a tree branch at dusk",
    "scattered autumn leaves on a park bench",
    "a stack of old letters tied with ribbon",
    "a glass bottle with a ship inside",
    "a row of paint jars with brushes in a studio",
    "a ceramic teapot with steam rising",
    "a pocket watch on a chain draped over books",
    "wildflowers in a mason jar on a kitchen table",
    "a telescope pointed at the night sky",
    "a loaf of crusty bread on a cutting board",
    "a pair of reading glasses on an open newspaper",
    "a birdcage with the door open and no bird inside",
    "a collection of seashells on a sandy surface",
    "a coffee cup leaving a ring stain on a book",
    "a set of watercolor paints with a half-finished painting",
    "a globe on a desk with warm lamp light",
    "a fishing rod leaning against a wooden dock post",
    "a chess board mid-game on an old table",
    "a cello leaning against a wall in an empty room",
    "a woven basket full of yarn and knitting needles",
    "a clay pot with herbs growing on a sunny ledge",
    "a bicycle leaning against a stone wall",
    "an ink bottle and quill on parchment paper",
    "a pair of ballet slippers on a wooden floor",
    "a bowl of ramen with steam rising",
    "a camera on a tripod facing a mountain view",
    "a skull with a candle on a dusty shelf",
    "a snow globe with a tiny village inside",
    "a record player with a vinyl spinning",
    "a watering can next to potted flowers",
    "a jar full of fireflies glowing softly",
    "a sword stuck in a mossy stone",
    "a hammock strung between two palm trees",
    "an open music box with a tiny ballerina",
    "a well-worn leather journal with a pen",
    "a ceramic bowl of fresh strawberries",
    "a broken pocket watch on cobblestones",
    "a ship in a bottle on a mantelpiece",
    "a bundle of dried lavender on a wooden table",

    # =========================================================================
    # 6. WEATHER & ATMOSPHERIC (50)
    # =========================================================================
    "a thunderstorm over a dark ocean with lightning",
    "a rainbow arcing over a green valley after rain",
    "dense fog rolling through a city at dawn",
    "a blizzard obscuring a lone cabin in the woods",
    "golden hour light casting long shadows on a field",
    "a dust storm approaching across a flat desert",
    "rain falling on a quiet Japanese garden",
    "a tornado funnel cloud over a flat prairie",
    "morning frost on a spider web in a garden",
    "a double rainbow over a waterfall",
    "haze rising from a lake on a warm morning",
    "a sun halo visible through high clouds",
    "heavy rain on a tin roof with splashing drops",
    "a dramatic cloudburst with rays of light breaking through",
    "snow falling gently on a sleeping village",
    "a sunset with bands of orange purple and red",
    "wind bending trees sideways near a coastline",
    "aurora borealis over a frozen lake",
    "a moody overcast sky over a grey sea",
    "dappled sunlight through dense forest canopy",
    "ice forming on tree branches in freezing rain",
    "a monsoon rain flooding a tropical street",
    "wispy cirrus clouds at extreme altitude",
    "a fog bank engulfing a suspension bridge",
    "a heat shimmer rising from hot asphalt",
    "fresh snow on rooftops in a quiet town",
    "a storm cell with rotating clouds over farmland",
    "mist rising from a waterfall in morning light",
    "a clear starry sky over a dark desert",
    "rays of light streaming through a break in storm clouds",
    "frost patterns on a windowpane",
    "a sandstorm turning the sky orange-brown",
    "gentle rain on a pond with expanding ripples",
    "sun dogs flanking the winter sun",
    "a cold dawn with breath visible in the air",
    "cumulus clouds towering over a green landscape",
    "a hailstorm with ice bouncing off a stone path",
    "fog lifting from a river at sunrise",
    "moonlight through thin clouds over a field",
    "a sudden downpour in a sunlit park",
    "rime ice coating every surface in a frozen forest",
    "a red sky at dawn over calm water",
    "sleet falling on a grey city street",
    "mammatus clouds hanging under a dark sky",
    "warm light glowing through autumn fog",
    "a cold front arriving with dark wall clouds",
    "dewdrops on morning grass in low sunlight",
    "a calm after a storm with wet glistening surfaces",
    "volcanic ash falling like grey snow on a town",
    "a spectacular cloud inversion filling a mountain valley",

    # =========================================================================
    # 7. FANTASY & SURREAL (50)
    # =========================================================================
    "a floating island with a waterfall pouring into the void",
    "a giant ancient tree with a city built in its branches",
    "a staircase spiraling into the clouds",
    "a door standing alone in an empty field opening to light",
    "a crystal cave with glowing formations",
    "a bridge made of roots spanning a bottomless chasm",
    "a clockwork city with gears turning in the sky",
    "an underwater library with fish swimming between shelves",
    "a field of giant glowing mushrooms at night",
    "a castle built upside-down hanging from a floating rock",
    "a train track running straight through the clouds",
    "a forest where the trees are made of glass",
    "a village inside a giant seashell",
    "a lake that reflects a different sky than the one above",
    "a massive stone hand emerging from the earth",
    "a ship sailing through a sea of clouds",
    "a room where gravity pulls in multiple directions",
    "a garden growing in zero gravity with floating soil",
    "a canyon with rivers of glowing light instead of water",
    "a tree with leaves made of tiny flames",
    "a pathway made of floating stepping stones over darkness",
    "a cathedral built entirely of ice on a frozen sea",
    "a forest of enormous swords stuck in the ground",
    "a mirror that shows a different world inside it",
    "an enormous sleeping dragon curled around a mountain",
    "a city built on the back of a colossal turtle",
    "a clock tower where time flows backward",
    "a waterfall flowing upward into the sky",
    "a lantern that illuminates things that aren't there",
    "a field of flowers that glow different colors at night",
    "an ancient portal made of intertwined stone serpents",
    "a sky with two moons and rings like Saturn",
    "a desert where the sand is made of tiny crystals",
    "a library tower that stretches infinitely upward",
    "a path through a forest of giant feathers",
    "a boat on a river that flows through the air",
    "a town where every building is a different impossible shape",
    "an hourglass the size of a building in a town square",
    "a rain of glowing petals falling from a dark sky",
    "a bridge between two mountain peaks wrapped in clouds",
    "a hollow mountain with a sun-like orb inside",
    "a maze of mirrors reflecting infinite corridors",
    "a whale swimming through clouds above a sleeping city",
    "a throne room inside a hollow glacier",
    "a garden where stone statues are slowly coming to life",
    "a river that splits into branching streams of light",
    "a sky full of floating paper lanterns at night",
    "a giant compass rose embedded in a stone courtyard",
    "a tree growing from the spine of a massive old book",
    "a valley where shadows move independently of their owners",

    # =========================================================================
    # 8. INTERIOR SCENES (50)
    # =========================================================================
    "a cozy reading nook with pillows and warm lamp light",
    "a sunlit art studio with canvases everywhere",
    "an old apothecary with jars lining the walls",
    "a dimly lit tavern with a roaring fireplace",
    "a Japanese tea room with tatami mats and a low table",
    "a cluttered workshop with tools and half-built inventions",
    "a grand ballroom with chandeliers and polished floors",
    "a cramped submarine interior with pipes and gauges",
    "a warm bakery kitchen with bread cooling on racks",
    "an attic room with a skylight and dust motes in the air",
    "a vintage train compartment with plush red seats",
    "a wizard's study with floating books and star charts",
    "a plant-filled sunroom with wicker furniture",
    "an empty theater stage with a single spotlight",
    "a rustic cabin kitchen with copper pots and herbs hanging",
    "a modern minimalist apartment with floor-to-ceiling windows",
    "a candlelit wine cellar with oak barrels",
    "a children's bedroom with glow-in-the-dark stars on the ceiling",
    "a space station observation deck with Earth visible",
    "a Victorian parlor with heavy curtains and ornate furniture",
    "a recording studio with microphones and mixing boards",
    "a greenhouse with tropical plants and humid air",
    "a ship captain's quarters with maps and a sextant",
    "a small chapel with light through stained glass",
    "a laundromat with warm fluorescent light and spinning machines",
    "a clockmaker's workshop with hundreds of ticking clocks",
    "a cozy mountain lodge with wooden beams and a fire",
    "a painter's palette table with mixed colors and brushes",
    "an old bookshop with narrow aisles and stacked volumes",
    "a diner at night with neon signs and vinyl booths",
    "a meditation room with a single candle and cushion",
    "a vintage pharmacy with labeled glass bottles",
    "a potter's studio with a spinning wheel and clay",
    "an astronomer's dome with a large telescope",
    "a barber shop with classic chairs and mirrors",
    "a kitchen with a window looking out on a garden",
    "a museum hall with a single sculpture in the center",
    "a dentist's waiting room from the 1960s",
    "a radio station booth with a glowing on-air sign",
    "a dimly lit jazz bar with a piano on stage",
    "a florist shop with buckets of fresh flowers",
    "a tailor's workshop with fabric bolts and mannequins",
    "a cozy tent interior with rugs and lantern light",
    "a scientist's lab with bubbling flasks and notebooks",
    "a yarn shop with colorful skeins filling the walls",
    "a vintage camera shop with old equipment displayed",
    "a calligraphy studio with ink and brushes",
    "a candle maker's workshop with wax and molds",
    "a map room with charts pinned to every surface",
    "a conservatory with a grand piano and tall windows",

    # =========================================================================
    # 9. ABSTRACT & COMPOSITIONAL (50)
    # =========================================================================
    "flowing organic shapes in warm amber tones",
    "geometric crystal formations catching prismatic light",
    "concentric ripples expanding across a calm surface",
    "swirling nebula colors blending into darkness",
    "layered translucent sheets of color overlapping",
    "a fractal pattern emerging from a single point",
    "ink dropped into water creating flowing tendrils",
    "a gradient from deep ocean blue to pale sky",
    "tangled threads of gold on a dark background",
    "soft bokeh circles of light floating in space",
    "a symmetrical mandala pattern in earth tones",
    "cracked dried earth with a single green sprout",
    "overlapping silhouettes in different muted colors",
    "a single drop of water creating a crown splash",
    "woven texture patterns in natural fiber colors",
    "smoke trails curling and intertwining in still air",
    "a mosaic of broken tiles in blue and green",
    "light refracting through a prism into a spectrum",
    "peeling paint layers revealing colors underneath",
    "a topographic map rendered as undulating terrain",
    "sand patterns created by wind from above",
    "tree rings seen in cross-section with varied tones",
    "aurora-like curtains of color in a dark void",
    "the fibonacci spiral rendered in natural elements",
    "scattered autumn leaves creating a natural collage",
    "ice crystal formations under a microscope",
    "water flowing over smooth river stones",
    "a color field with soft edges bleeding into each other",
    "marble veining patterns in grey and white",
    "shadows of leaves creating patterns on a wall",
    "overlapping circles in complementary colors",
    "a single feather with detailed barb structure",
    "erosion patterns in layered sedimentary rock",
    "morning dew forming beads on a spider silk strand",
    "cross-section of a nautilus shell",
    "weathered wood grain texture in warm browns",
    "cloud formations that suggest faces or figures",
    "rippled reflection of trees in disturbed water",
    "oxidized copper surface with blue-green patina",
    "a field of identical shapes receding into perspective",
    "interleaving waves of warm and cool colors",
    "crumpled paper with dramatic side lighting",
    "the texture of old stone with lichen growth",
    "parallel lines of light and shadow from blinds",
    "bubbles of varying sizes catching rainbow light",
    "a seashell spiral viewed from the opening",
    "terraced mineral deposits in a hot spring",
    "the negative space between tree branches",
    "overlapping transparent wings of a dragonfly",
    "concentric growth rings of a sliced agate",

    # =========================================================================
    # 10. SEASONAL & TIME-OF-DAY (50)
    # =========================================================================
    "a village in the first light of a spring morning",
    "a summer afternoon by a lazy river with tall grass",
    "an autumn park with red and gold leaves falling",
    "a winter evening with snow falling on a lit window",
    "dawn breaking over a fishing harbor",
    "high noon in a dusty western town",
    "the blue hour over a quiet suburban street",
    "midnight in a garden with only moonlight",
    "a spring garden with crocuses pushing through snow",
    "a hot summer day with heat haze over a road",
    "an autumn harvest scene with pumpkins and corn",
    "a frozen winter stream with icicles on branches",
    "the first rays of sunrise hitting a mountain peak",
    "a long summer evening with fireflies appearing",
    "an autumn bonfire with sparks rising into darkness",
    "a winter morning with fresh untouched snow",
    "a spring rain shower with sun still shining",
    "a summer thunderstorm approaching over a lake",
    "the last autumn leaf falling from a bare tree",
    "a cozy winter scene through a frosted window",
    "cherry blossoms falling in a spring breeze",
    "a midsummer night with a low orange moon",
    "an October country road with trees in peak color",
    "a December street with holiday lights reflecting on wet pavement",
    "a misty spring meadow at first light",
    "a scorching summer desert under a white sun",
    "an autumn woodland path with a carpet of leaves",
    "a deep winter night with stars over a snowy field",
    "tulips opening in a spring morning garden",
    "a lazy summer hammock scene by a pond",
    "fog in an autumn cemetery with old headstones",
    "a winter cabin with smoke rising from the chimney",
    "spring flooding in a river valley",
    "a summer night market with string lights",
    "an autumn vineyard with golden afternoon light",
    "a winter ice-skating pond at twilight",
    "a spring waterfall at full flow from snowmelt",
    "a summer camping scene by a mountain lake",
    "an autumn rainstorm through a cafe window",
    "a winter landscape in flat grey overcast light",
    "a bright spring day in a city park",
    "a summer beach at golden hour with long shadows",
    "an autumn farmers market with root vegetables",
    "the first frost of winter on a garden gate",
    "a spring morning with birds at a feeder",
    "a humid summer jungle with afternoon storm light",
    "an autumn sunset behind a bare oak tree",
    "a snowy winter bridge with lanterns",
    "a spring creek with clear water over pebbles",
    "a summer stargazing scene on a hilltop blanket",
]

# =============================================================================
# SUBSET SELECTION
# =============================================================================
# The full list above is 10 category blocks of 50 prompts each. For a small
# run we take PER_CATEGORY prompts evenly spaced inside each block, so the
# subset still spans every content type instead of only the first few blocks.

ALL_PROMPTS = BASE_PROMPTS
BLOCK_SIZE = 50
CATEGORY_NAMES = [
    "natural_environments",
    "urban_architecture",
    "characters_figures",
    "animals_creatures",
    "still_life_objects",
    "weather_atmospheric",
    "fantasy_surreal",
    "interior_scenes",
    "abstract_compositional",
    "seasonal_time_of_day",
]
assert len(ALL_PROMPTS) == BLOCK_SIZE * len(CATEGORY_NAMES)

PER_CATEGORY = 2  # 2 x 10 categories = 20 prompts = 20 images per model


def _subset(per_category):
    prompts, categories = [], {}
    for block, name in enumerate(CATEGORY_NAMES):
        start = len(prompts)
        stride = BLOCK_SIZE // per_category
        for k in range(per_category):
            prompts.append(ALL_PROMPTS[block * BLOCK_SIZE + k * stride])
        categories[name] = (start, len(prompts))
    return prompts, categories


BASE_PROMPTS, CATEGORIES = _subset(PER_CATEGORY)

assert len(BASE_PROMPTS) == PER_CATEGORY * len(CATEGORY_NAMES)


def _restyle(prompt, style):
    """Swap the leading article for '<article> <style> ...' so the two arms
    differ only by the style phrase. Note: a plain .lstrip('a ') here would
    eat leading vowels ('a ancient tower' -> 'ncient tower')."""
    if prompt.startswith("a "):
        body = prompt[2:]
    elif prompt.startswith("an "):
        body = prompt[3:]
    else:
        body = prompt
    article = "an" if style[0].lower() in "aeiou" else "a"
    return f"{article} {style} {body}"


def get_lora_prompts():
    """Prompts with the <s0><s1> trigger, for LoRA generation."""
    return [_restyle(p, "<s0><s1> style") for p in BASE_PROMPTS]


def get_vanilla_prompts():
    """Prompts with a neutral style prefix, for the vanilla SDXL baseline."""
    return [_restyle(p, "painterly") for p in BASE_PROMPTS]


def get_clip_prompts():
    """Content-only prompts. Both arms are scored against these so CLIP
    measures prompt adherence, not the style words (and so the untrained
    <s0><s1> tokens don't pollute the LoRA arm's text embedding)."""
    return list(BASE_PROMPTS)
```

```python
"""
generate_images.py — Generate one image per prompt for vanilla SDXL and LoRA-finetuned SDXL.

The prompt count is whatever prompts.py exposes as BASE_PROMPTS (currently 20).

Designed for Kaggle free tier (dual T4 GPUs, ~15GB VRAM each).
Features:
  - Resume support: skips already-generated images on restart
  - Memory-efficient: VAE slicing/tiling, explicit cache clearing between phases
  - Deterministic: same seed per prompt index for fair comparison
  - Batch logging: progress bar with ETA

Usage on Kaggle:
    python generate_images.py
"""

import torch
import os
import time
import json
from diffusers import StableDiffusionXLPipeline
from PIL import Image

from prompts import get_lora_prompts, get_vanilla_prompts, BASE_PROMPTS, CATEGORIES

DEVICE = "cuda" if torch.cuda.is_available() else "cpu"

# =============================================================================
# CONFIG — adjust paths for your Kaggle environment
# =============================================================================
BASE_MODEL = "/kaggle/input/models/stabilityai/stable-diffusion-xl/pytorch/base-1-0/1"
LORA_PATH = "/kaggle/input/datasets/bimukti/lora-neva-1000"
EMBED_PATH = "/kaggle/input/datasets/bimukti/embeddings"

OUTPUT_ROOT = "/kaggle/working/eval_outputs"
VANILLA_DIR = os.path.join(OUTPUT_ROOT, "vanilla")
LORA_DIR = os.path.join(OUTPUT_ROOT, "lora")
META_PATH = os.path.join(OUTPUT_ROOT, "generation_meta.json")

os.makedirs(VANILLA_DIR, exist_ok=True)
os.makedirs(LORA_DIR, exist_ok=True)

# Generation params
BASE_SEED = 42
CFG_SCALE = 7.0
NUM_STEPS = 30
WIDTH = 1024
HEIGHT = 1024

NEGATIVE_PROMPT = (
    "photorealistic, 3D render, sharp edges, cartoon, anime, "
    "low quality, blurry, text, watermark, UI elements"
)


def image_path(output_dir, idx):
    return os.path.join(output_dir, f"{idx:04d}.png")


def count_existing(output_dir, total):
    """Count how many images already exist (for resume)."""
    return sum(1 for i in range(total) if os.path.exists(image_path(output_dir, i)))


def generate_batch(pipe, prompts, output_dir, label=""):
    """Generate images for all prompts, skipping those already on disk."""
    total = len(prompts)
    existing = count_existing(output_dir, total)
    if existing == total:
        print(f"[{label}] All {total} images already exist. Skipping.")
        return
    print(f"[{label}] {existing}/{total} already exist. Generating remaining...")

    start_time = time.time()
    generated = 0

    for i, prompt in enumerate(prompts):
        out_path = image_path(output_dir, i)
        if os.path.exists(out_path):
            continue

        seed = BASE_SEED + i
        generator = torch.Generator(device=DEVICE).manual_seed(seed)

        image = pipe(
            prompt=prompt,
            negative_prompt=NEGATIVE_PROMPT,
            guidance_scale=CFG_SCALE,
            num_inference_steps=NUM_STEPS,
            width=WIDTH,
            height=HEIGHT,
            generator=generator,
        ).images[0]

        image.save(out_path)
        generated += 1

        # Progress logging every 10 images (or every image on tiny runs)
        if generated % max(1, min(10, total // 4)) == 0:
            elapsed = time.time() - start_time
            rate = generated / elapsed
            remaining = (total - existing - generated) / rate if rate > 0 else 0
            print(
                f"  [{label}] {existing + generated}/{total} done | "
                f"{rate:.2f} img/s | ~{remaining/60:.1f} min remaining"
            )

        # Memory management every 50 images
        if generated % 50 == 0 and DEVICE == "cuda":
            torch.cuda.empty_cache()

    elapsed = time.time() - start_time
    print(f"[{label}] Done. Generated {generated} new images in {elapsed/60:.1f} min.")


def load_embeddings(pipe, embed_path):
    """Load textual inversion embeddings for <s0><s1> tokens."""
    embed_file = None
    for f in os.listdir(embed_path):
        if f.endswith(".safetensors"):
            embed_file = os.path.join(embed_path, f)
            break

    if embed_file is None:
        print("WARNING: No embeddings .safetensors file found!")
        return False

    from safetensors.torch import load_file
    state_dict = load_file(embed_file)
    pipe.load_textual_inversion(
        state_dict["clip_l"],
        token=["<s0>", "<s1>"],
        text_encoder=pipe.text_encoder,
        tokenizer=pipe.tokenizer,
    )
    pipe.load_textual_inversion(
        state_dict["clip_g"],
        token=["<s0>", "<s1>"],
        text_encoder=pipe.text_encoder_2,
        tokenizer=pipe.tokenizer_2,
    )
    print(f"Loaded embeddings from {embed_file}")
    return True


def main():
    vanilla_prompts = get_vanilla_prompts()
    lora_prompts = get_lora_prompts()
    total = len(BASE_PROMPTS)

    print(f"Total prompts: {total}")
    print(f"Vanilla output: {VANILLA_DIR}")
    print(f"LoRA output:    {LORA_DIR}")
    print()

    # -----------------------------------------------------------------
    # PHASE 1: Vanilla SDXL
    # -----------------------------------------------------------------
    vanilla_needed = total - count_existing(VANILLA_DIR, total)
    if vanilla_needed > 0:
        print("=" * 60)
        print("PHASE 1: Vanilla SDXL generation")
        print("=" * 60)
        pipe = StableDiffusionXLPipeline.from_pretrained(
            BASE_MODEL,
            torch_dtype=torch.float16,
            use_safetensors=True,
            variant="fp16",
        ).to(DEVICE)
        pipe.enable_vae_slicing()
        pipe.enable_vae_tiling()

        generate_batch(pipe, vanilla_prompts, VANILLA_DIR, label="Vanilla")

        # Free memory before LoRA phase
        del pipe
        if DEVICE == "cuda":
            torch.cuda.empty_cache()
    else:
        print("PHASE 1: All vanilla images exist. Skipping.\n")

    # -----------------------------------------------------------------
    # PHASE 2: LoRA SDXL
    # -----------------------------------------------------------------
    lora_needed = total - count_existing(LORA_DIR, total)
    if lora_needed > 0:
        print("=" * 60)
        print("PHASE 2: LoRA SDXL generation")
        print("=" * 60)
        pipe = StableDiffusionXLPipeline.from_pretrained(
            BASE_MODEL,
            torch_dtype=torch.float16,
            use_safetensors=True,
            variant="fp16",
        ).to(DEVICE)
        pipe.enable_vae_slicing()
        pipe.enable_vae_tiling()

        # Load LoRA weights + embeddings
        pipe.load_lora_weights(LORA_PATH)
        load_embeddings(pipe, EMBED_PATH)

        generate_batch(pipe, lora_prompts, LORA_DIR, label="LoRA")

        del pipe
        if DEVICE == "cuda":
            torch.cuda.empty_cache()
    else:
        print("PHASE 2: All LoRA images exist. Skipping.\n")

    # -----------------------------------------------------------------
    # Save metadata
    # -----------------------------------------------------------------
    meta = {
        "total_prompts": total,
        "base_seed": BASE_SEED,
        "cfg_scale": CFG_SCALE,
        "num_steps": NUM_STEPS,
        "resolution": f"{WIDTH}x{HEIGHT}",
        "negative_prompt": NEGATIVE_PROMPT,
        "categories": CATEGORIES,
        "prompts": {
            f"{i:04d}": {"base": b, "vanilla": v, "lora": l}
            for i, (b, v, l) in enumerate(zip(BASE_PROMPTS, vanilla_prompts, lora_prompts))
        },
        "vanilla_count": count_existing(VANILLA_DIR, total),
        "lora_count": count_existing(LORA_DIR, total),
    }
    with open(META_PATH, "w") as f:
        json.dump(meta, f, indent=2)
    print(f"\nMetadata saved to {META_PATH}")
    print(f"Vanilla: {meta['vanilla_count']}/{total}")
    print(f"LoRA:    {meta['lora_count']}/{total}")
    print("\nGeneration complete! Run evaluate.py next.")


if __name__ == "__main__":
    main()
```

```python
"""
evaluate.py — Compute evaluation metrics for vanilla vs LoRA style transfer.

Metrics:
  1. Gram Matrix Style Loss — style similarity to training data (primary)
  2. CLIP Score — text-image alignment, scored on content-only prompts
  3. LPIPS — perceptual distance between paired images (same prompt+seed)
  4. FID — distribution distance; only meaningful with >= ~200 images per set

Runs both as a script and inside a notebook cell:
    python evaluate.py --skip-fid
    # or, in a notebook:  ARGS = Args(skip_fid=True); evaluate(ARGS)
"""

import torch
import torch.nn as nn
import torchvision.transforms as T
import torchvision.models as models
import numpy as np
import json
import os
import sys
import argparse
from dataclasses import dataclass
from PIL import Image
from pathlib import Path

# =============================================================================
# CONFIG
# =============================================================================
OUTPUT_ROOT = "/kaggle/working/eval_outputs"
VANILLA_DIR = os.path.join(OUTPUT_ROOT, "vanilla")
LORA_DIR = os.path.join(OUTPUT_ROOT, "lora")
TRAINING_DATA_DIR = "/kaggle/input/datasets/bimukti/neva-images"  # adjust to your training images
RESULTS_PATH = os.path.join(OUTPUT_ROOT, "eval_results.json")

DEVICE = "cuda" if torch.cuda.is_available() else "cpu"

# FID is heavily biased on small samples; below this it is not reported.
FID_MIN_IMAGES = 200

# prompts.py is written by the first cell into the current working directory.
sys.path.insert(0, os.getcwd())
from prompts import BASE_PROMPTS, CATEGORIES, get_clip_prompts


@dataclass
class Args:
    """Notebook-friendly stand-in for argparse results."""
    skip_fid: bool = False
    category: str = None


def parse_args(argv=None):
    """Parse CLI args when run as a script; return defaults inside a kernel.

    argparse would otherwise choke on the kernel's own -f <connection file>
    and raise SystemExit(2).
    """
    in_notebook = "ipykernel" in sys.modules or "google.colab" in sys.modules
    if in_notebook and argv is None:
        return Args()
    parser = argparse.ArgumentParser(description="Evaluate vanilla vs LoRA SDXL")
    parser.add_argument("--skip-fid", action="store_true", help="Skip FID computation")
    parser.add_argument("--category", type=str, default=None, help="Evaluate single category")
    ns = parser.parse_args(argv)
    return Args(skip_fid=ns.skip_fid, category=ns.category)


# =============================================================================
# UTILITIES
# =============================================================================
def load_images_for_indices(directory, indices):
    """Load {idx:04d}.png for each index. Returns (images, kept_indices) so the
    caller can keep prompts aligned when some renders are missing."""
    images, kept = [], []
    for i in indices:
        path = os.path.join(directory, f"{i:04d}.png")
        if os.path.exists(path):
            images.append(Image.open(path).convert("RGB"))
            kept.append(i)
    return images, kept


def load_reference_images(root):
    """Load training images from a directory tree (any common image format)."""
    exts = {".png", ".jpg", ".jpeg", ".webp", ".bmp"}
    files = sorted(f for f in Path(root).rglob("*") if f.is_file() and f.suffix.lower() in exts)
    return [Image.open(str(f)).convert("RGB") for f in files]


def free_cuda():
    if DEVICE == "cuda":
        torch.cuda.empty_cache()


# =============================================================================
# METRIC 1: FID (using clean-fid)
# =============================================================================
def compute_fid(dir1, dir2):
    """Compute FID between two directories of images using clean-fid."""
    try:
        from cleanfid import fid
        return fid.compute_fid(dir1, dir2, device=torch.device(DEVICE))
    except ImportError:
        print("clean-fid not installed. Install with: pip install clean-fid")
        return None


# =============================================================================
# METRIC 2: CLIP Score (text-image alignment)
# =============================================================================
def compute_clip_scores(images, prompts):
    """Per-image CLIP cosine similarity against the matching prompt."""
    assert len(images) == len(prompts), "images and prompts must be aligned"
    try:
        import open_clip
    except ImportError:
        print("open_clip not installed. Install with: pip install open_clip_torch")
        return None

    model, _, preprocess = open_clip.create_model_and_transforms(
        "ViT-B-32", pretrained="laion2b_s34b_b79k"
    )
    tokenizer = open_clip.get_tokenizer("ViT-B-32")
    model = model.to(DEVICE).eval()

    scores = []
    batch_size = 32

    for i in range(0, len(images), batch_size):
        batch_imgs = images[i:i + batch_size]
        batch_prompts = prompts[i:i + batch_size]

        img_tensors = torch.stack([preprocess(img) for img in batch_imgs]).to(DEVICE)
        text_tokens = tokenizer(batch_prompts).to(DEVICE)

        with torch.no_grad():
            img_features = model.encode_image(img_tensors)
            text_features = model.encode_text(text_tokens)
            img_features = img_features / img_features.norm(dim=-1, keepdim=True)
            text_features = text_features / text_features.norm(dim=-1, keepdim=True)
            similarity = (img_features * text_features).sum(dim=-1)
            scores.extend(similarity.float().cpu().numpy().tolist())

    del model
    free_cuda()
    return np.array(scores)


# =============================================================================
# METRIC 3: LPIPS (perceptual distance between vanilla & LoRA per prompt)
# =============================================================================
def compute_lpips_scores(images_a, images_b):
    """Per-pair LPIPS between two aligned sets of images."""
    assert len(images_a) == len(images_b), "LPIPS needs aligned pairs"
    try:
        import lpips
    except ImportError:
        print("lpips not installed. Install with: pip install lpips")
        return None

    loss_fn = lpips.LPIPS(net="alex").to(DEVICE).eval()

    transform = T.Compose([
        T.Resize((256, 256)),
        T.ToTensor(),
        T.Normalize(mean=[0.5, 0.5, 0.5], std=[0.5, 0.5, 0.5]),  # LPIPS expects [-1, 1]
    ])

    scores = []
    batch_size = 16

    for i in range(0, len(images_a), batch_size):
        batch_a = torch.stack([transform(img) for img in images_a[i:i + batch_size]]).to(DEVICE)
        batch_b = torch.stack([transform(img) for img in images_b[i:i + batch_size]]).to(DEVICE)

        with torch.no_grad():
            d = loss_fn(batch_a, batch_b)
        # d is (B, 1, 1, 1); reshape (not squeeze) so a batch of 1 stays 1-D.
        scores.extend(d.reshape(-1).float().cpu().numpy().tolist())

    del loss_fn
    free_cuda()
    return np.array(scores)


# =============================================================================
# METRIC 4: Gram Matrix Style Loss
# =============================================================================
class GramStyleExtractor(nn.Module):
    """Extract Gram matrices from VGG19 feature layers for style comparison."""

    def __init__(self):
        super().__init__()
        vgg = models.vgg19(weights=models.VGG19_Weights.IMAGENET1K_V1).features
        # Style layers: conv1_1, conv2_1, conv3_1, conv4_1, conv5_1
        self.slices = nn.ModuleList([
            vgg[:2],    # conv1_1
            vgg[2:7],   # conv2_1
            vgg[7:12],  # conv3_1
            vgg[12:21], # conv4_1
            vgg[21:30], # conv5_1
        ])
        for p in self.parameters():
            p.requires_grad = False

    def gram_matrix(self, x):
        b, c, h, w = x.shape
        features = x.view(b, c, h * w)
        gram = torch.bmm(features, features.transpose(1, 2))
        return gram / (c * h * w)

    def forward(self, x):
        grams = []
        for s in self.slices:
            x = s(x)
            grams.append(self.gram_matrix(x))
        return grams


_STYLE_TRANSFORM = T.Compose([
    T.Resize((256, 256)),
    T.ToTensor(),
    T.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),
])


def reference_gram_stats(extractor, reference_images, batch_size=16):
    """Mean Gram matrix per style layer over the reference (training) images."""
    ref_sum, ref_count = None, 0
    for i in range(0, len(reference_images), batch_size):
        batch = torch.stack(
            [_STYLE_TRANSFORM(img) for img in reference_images[i:i + batch_size]]
        ).to(DEVICE)
        with torch.no_grad():
            grams = extractor(batch)
        if ref_sum is None:
            ref_sum = [g.sum(dim=0) for g in grams]
        else:
            for j, g in enumerate(grams):
                ref_sum[j] += g.sum(dim=0)
        ref_count += batch.shape[0]
    return [g / ref_count for g in ref_sum]


def gram_style_loss(extractor, ref_grams_mean, images, batch_size=16):
    """Per-image style distance to the reference style. Each layer's MSE is
    normalised by that layer's reference Gram energy, otherwise the single
    largest-magnitude layer decides the whole score."""
    layer_scale = [float(torch.mean(g ** 2).item()) + 1e-12 for g in ref_grams_mean]
    losses = []
    for i in range(0, len(images), batch_size):
        batch = torch.stack(
            [_STYLE_TRANSFORM(img) for img in images[i:i + batch_size]]
        ).to(DEVICE)
        with torch.no_grad():
            grams = extractor(batch)
        for img_idx in range(batch.shape[0]):
            loss = 0.0
            for layer_idx, (gen_gram, ref_gram) in enumerate(zip(grams, ref_grams_mean)):
                diff = gen_gram[img_idx] - ref_gram
                loss += torch.mean(diff ** 2).item() / layer_scale[layer_idx]
            losses.append(loss / len(ref_grams_mean))
    return np.array(losses)


# =============================================================================
# MAIN EVALUATION
# =============================================================================
def evaluate(args):
    results = {"n_prompts_requested": len(BASE_PROMPTS)}

    # Determine which indices to evaluate
    if args.category:
        if args.category not in CATEGORIES:
            print(f"Unknown category: {args.category}")
            print(f"Available: {list(CATEGORIES.keys())}")
            return
        start, end = CATEGORIES[args.category]
        indices = list(range(start, end))
        print(f"Evaluating category: {args.category} (indices {start}-{end})")
    else:
        indices = list(range(len(BASE_PROMPTS)))
        print(f"Evaluating all {len(BASE_PROMPTS)} prompts")

    # Load images, keeping only indices rendered by BOTH models so every
    # metric compares the same prompts and LPIPS pairs line up.
    print("\nLoading images...")
    _, vanilla_idx = load_images_for_indices(VANILLA_DIR, indices)
    _, lora_idx = load_images_for_indices(LORA_DIR, indices)
    paired = [i for i in indices if i in set(vanilla_idx) & set(lora_idx)]
    missing = len(indices) - len(paired)
    if missing:
        print(f"  WARNING: {missing} prompt(s) missing from one or both models; dropped.")
    if not paired:
        print("ERROR: No paired images found. Run generate_images.py first.")
        return

    vanilla_images, _ = load_images_for_indices(VANILLA_DIR, paired)
    lora_images, _ = load_images_for_indices(LORA_DIR, paired)
    print(f"  Vanilla: {len(vanilla_images)} images")
    print(f"  LoRA:    {len(lora_images)} images")
    results["n_pairs_evaluated"] = len(paired)

    clip_prompts_full = get_clip_prompts()
    clip_prompts = [clip_prompts_full[i] for i in paired]

    # ------------------------------------------------------------------
    # Gram Matrix Style Loss — the metric that actually answers
    # "did the LoRA move toward the target style?"
    # ------------------------------------------------------------------
    print("\n--- Gram Matrix Style Loss (vs training data) ---")
    if os.path.isdir(TRAINING_DATA_DIR):
        ref_images = load_reference_images(TRAINING_DATA_DIR)
        if ref_images:
            print(f"  Loaded {len(ref_images)} reference (training) images")
            extractor = GramStyleExtractor().to(DEVICE).eval()
            print("  Computing reference style statistics...")
            ref_grams_mean = reference_gram_stats(extractor, ref_images)
            print("  Computing vanilla style loss...")
            vanilla_style = gram_style_loss(extractor, ref_grams_mean, vanilla_images)
            print("  Computing LoRA style loss...")
            lora_style = gram_style_loss(extractor, ref_grams_mean, lora_images)
            del extractor
            free_cuda()

            improvement = (
                (np.mean(vanilla_style) - np.mean(lora_style)) / np.mean(vanilla_style) * 100
            )
            # Paired win rate: on how many prompts is LoRA closer to the style?
            win_rate = float(np.mean(lora_style < vanilla_style))

            results["gram_style_loss"] = {
                "vanilla_mean": float(np.mean(vanilla_style)),
                "vanilla_std": float(np.std(vanilla_style)),
                "lora_mean": float(np.mean(lora_style)),
                "lora_std": float(np.std(lora_style)),
                "improvement_pct": float(improvement),
                "lora_win_rate": win_rate,
                "reference_count": len(ref_images),
            }
            print(f"  Vanilla style loss: {np.mean(vanilla_style):.6f} ± {np.std(vanilla_style):.6f}")
            print(f"  LoRA style loss:    {np.mean(lora_style):.6f} ± {np.std(lora_style):.6f}")
            print(f"  (Lower = closer to training data style)")
            print(f"  Style improvement: {improvement:+.1f}% (positive = LoRA closer to target)")
            print(f"  LoRA wins on {win_rate * 100:.0f}% of paired prompts")
        else:
            print(f"  WARNING: No images found in {TRAINING_DATA_DIR}")
    else:
        print(f"  WARNING: Training data directory not found: {TRAINING_DATA_DIR}")
        print(f"  Update TRAINING_DATA_DIR to point to your training images.")

    # ------------------------------------------------------------------
    # CLIP Score — both arms scored against the SAME content-only prompt,
    # so this measures prompt adherence rather than the style wording.
    # ------------------------------------------------------------------
    print("\n--- CLIP Score (prompt adherence, content-only prompts) ---")
    vanilla_clip = compute_clip_scores(vanilla_images, clip_prompts)
    lora_clip = compute_clip_scores(lora_images, clip_prompts)

    if vanilla_clip is not None and lora_clip is not None:
        results["clip_score"] = {
            "vanilla_mean": float(np.mean(vanilla_clip)),
            "vanilla_std": float(np.std(vanilla_clip)),
            "lora_mean": float(np.mean(lora_clip)),
            "lora_std": float(np.std(lora_clip)),
            "delta": float(np.mean(lora_clip) - np.mean(vanilla_clip)),
        }
        print(f"  Vanilla CLIP: {np.mean(vanilla_clip):.4f} ± {np.std(vanilla_clip):.4f}")
        print(f"  LoRA CLIP:    {np.mean(lora_clip):.4f} ± {np.std(lora_clip):.4f}")
        print(f"  Delta:        {np.mean(lora_clip) - np.mean(vanilla_clip):+.4f} "
              f"(negative = LoRA lost prompt adherence)")

        # Per-category breakdown. Only informative when each category has a
        # meaningful number of prompts — at 2/category this is noise.
        pos = {idx: k for k, idx in enumerate(paired)}
        cat_clip = {}
        for cat_name, (start, end) in CATEGORIES.items():
            cat_pos = [pos[i] for i in range(start, end) if i in pos]
            if cat_pos:
                cat_clip[cat_name] = {
                    "n": len(cat_pos),
                    "vanilla_mean": float(np.mean(vanilla_clip[cat_pos])),
                    "lora_mean": float(np.mean(lora_clip[cat_pos])),
                }
        results["clip_per_category"] = cat_clip
        min_n = min((c["n"] for c in cat_clip.values()), default=0)
        if min_n >= 10:
            print("\n  Per-category CLIP scores:")
            for cat_name, c in cat_clip.items():
                print(f"    {cat_name:30s} V={c['vanilla_mean']:.4f}  L={c['lora_mean']:.4f}")
        else:
            print(f"\n  (Per-category breakdown saved to JSON but not printed: "
                  f"only {min_n} prompt(s) per category — too few to read.)")

    # ------------------------------------------------------------------
    # LPIPS (paired perceptual distance) — same prompt, same seed,
    # so this isolates how much the LoRA changed the render.
    # ------------------------------------------------------------------
    print("\n--- LPIPS (perceptual distance vanilla ↔ LoRA) ---")
    lpips_scores = compute_lpips_scores(vanilla_images, lora_images)
    if lpips_scores is not None:
        results["lpips"] = {
            "mean": float(np.mean(lpips_scores)),
            "std": float(np.std(lpips_scores)),
            "median": float(np.median(lpips_scores)),
        }
        print(f"  LPIPS mean:   {np.mean(lpips_scores):.4f} ± {np.std(lpips_scores):.4f}")
        print(f"  LPIPS median: {np.median(lpips_scores):.4f}")
        print(f"  (Higher = LoRA changed the image more; says nothing about direction)")

    # ------------------------------------------------------------------
    # FID — needs thousands of images to be trustworthy. Skipped by default
    # on small runs rather than reported as a misleading number.
    # ------------------------------------------------------------------
    print("\n--- FID (Fréchet Inception Distance) ---")
    if args.skip_fid:
        print("  Skipped (--skip-fid).")
    elif len(paired) < FID_MIN_IMAGES:
        print(f"  Skipped: only {len(paired)} images per set. FID is strongly biased "
              f"below ~{FID_MIN_IMAGES} and unusable below ~2000; the number would be "
              f"a function of sample size, not model quality.")
        results["fid_skipped_reason"] = f"n={len(paired)} < {FID_MIN_IMAGES}"
    else:
        for label, ref_dir in (("vanilla", VANILLA_DIR), ("lora", LORA_DIR)):
            if os.path.isdir(TRAINING_DATA_DIR):
                score = compute_fid(TRAINING_DATA_DIR, ref_dir)
                if score is not None:
                    print(f"  FID (training vs {label}): {score:.4f}")
                    results[f"fid_training_vs_{label}"] = score
        score = compute_fid(VANILLA_DIR, LORA_DIR)
        if score is not None:
            print(f"  FID (vanilla vs LoRA):    {score:.4f}  (separation only, not quality)")
            results["fid_vanilla_vs_lora"] = score

    # ------------------------------------------------------------------
    # Save results
    # ------------------------------------------------------------------
    suffix = f"_{args.category}" if args.category else ""
    save_path = RESULTS_PATH.replace(".json", f"{suffix}.json")
    os.makedirs(os.path.dirname(save_path), exist_ok=True)
    with open(save_path, "w") as f:
        json.dump(results, f, indent=2)
    print(f"\nResults saved to {save_path}")

    # ------------------------------------------------------------------
    # Summary
    # ------------------------------------------------------------------
    print("\n" + "=" * 60)
    print(f"EVALUATION SUMMARY  (n = {len(paired)} paired prompts)")
    print("=" * 60)
    if "gram_style_loss" in results:
        gs = results["gram_style_loss"]
        print(f"  Style Loss (Vanilla):  {gs['vanilla_mean']:.6f}")
        print(f"  Style Loss (LoRA):     {gs['lora_mean']:.6f}")
        print(f"  Style Improvement:     {gs['improvement_pct']:+.1f}%  "
              f"(LoRA wins {gs['lora_win_rate'] * 100:.0f}% of pairs)")
    if "clip_score" in results:
        cs = results["clip_score"]
        print(f"  CLIP Score (Vanilla):  {cs['vanilla_mean']:.4f} ± {cs['vanilla_std']:.4f}")
        print(f"  CLIP Score (LoRA):     {cs['lora_mean']:.4f} ± {cs['lora_std']:.4f}")
    if "lpips" in results:
        print(f"  LPIPS (V ↔ L):         {results['lpips']['mean']:.4f} ± {results['lpips']['std']:.4f}")
    if "fid_vanilla_vs_lora" in results:
        print(f"  FID (V vs L):          {results['fid_vanilla_vs_lora']:.2f}")
    if len(paired) < 50:
        print("\n  NOTE: at this sample size treat these as a smoke test, not a")
        print("  benchmark. Report the paired win rate and per-prompt spread,")
        print("  not the mean alone.")
    print("=" * 60)


evaluate(parse_args())
```
